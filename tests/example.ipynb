```python
import os
import sys
sys.path.append("../")
import mindkosh

mk_token = os.getenv("MK_SDK_TOKEN", "")
client = mindkosh.Client(
    token = mk_token
)
```

```python
tasks = client.task.get()
task1 = tasks[0]

public_callback_url = ""
task1.download_annotations(
    format = "coco",
    filename = "coco_annotations.zip",
    location = "./"
    public_url = public_callback_url
)
```

```python
frames = task1.frames
frames[0].visualize()
frame_annotations = frames[0].annotations()
```

```python
testset = mindkosh.TestSet()

testset.add(
    (frames[0], frames[2], frames[3])
)
testset.download_annotations(
    location="./"
)
```
